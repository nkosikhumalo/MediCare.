## Embeddings example

```python
EMBEDDING_MODELS = [
    "amazon.titan-embed-image-v1",
    "amazon.titan-embed-text-v2:0"
]

# prepend bedrock/ to each
for i in range(len(EMBEDDING_MODELS)):
    EMBEDDING_MODELS[i] = "bedrock/" + EMBEDDING_MODELS[i]
```

```python
for model in EMBEDDING_MODELS:
    try:
        embedding = client.embeddings.create(
            model=model,
            input=["The food was delicious and the waiter...", "The movie was terrible and the plot was..."],
            encoding_format="float"
        )
        print(f"Model: {model}, Embedding Length: {len(embedding.data[0].embedding)}")
    except Exception as e:
        print(f"Model: {model}, Error: {e}")
```
